## ***5. Hypothesis Testing***

Hypothesis 1

In this project, it is used to verify whether the historical stock price variables (Open, High, and Low) have a significant relationship with the Closing Price. Establishing these relationships helps justify the selection of features for the machine learning model.

Null Hypothesis (H₀)

There is no significant linear relationship between the Opening Price and the Closing Price.

Alternative Hypothesis (H₁)

There is a significant linear relationship between the Opening Price and the Closing Price.

````python
from scipy.stats import pearsonr

corr_open, p_open = pearsonr(data['Open'], data['Close'])

print("Correlation:", round(corr_open,4))
print("P-value:", p_open)
# If the p-value is less than 0.05 and the correlation coefficient is close to 1,
# it indicates that the Opening Price has a strong positive and statistically significant relationship with the Closing Price.
````

Hypothesis 2

Null Hypothesis (H₀)

There is no significant linear relationship between the Highest Price and the Closing Price.

Alternative Hypothesis (H₁)

There is a significant linear relationship between the Highest Price and the Closing Price

````python
corr_high, p_high = pearsonr(data['High'], data['Close'])

print("Correlation:", round(corr_high,4))
print("P-value:", p_high)
# If the p-value is less than 0.05, the Highest Price is statistically associated with the Closing Price
# and can be considered an important predictor.
````

Hypothesis 3

Null Hypothesis (H₀)

There is no significant linear relationship between the Lowest Price and the Closing Price.

Alternative Hypothesis (H₁)

There is a significant linear relationship between the Lowest Price and the Closing Price.

````python
corr_low, p_low = pearsonr(data['Low'], data['Close'])

print("Correlation:", round(corr_low,4))
print("P-value:", p_low)
# If the p-value is less than 0.05, the Lowest Price has a statistically significant relationship
# with the Closing Price.
````

## ***6. Feature Engineering & Data Pre-processing***

###Step 1: Feature Selection & Feature Engineering


The first step in machine learning is to identify the input (independent) variables and the target (dependent) variable.

Since the objective of this project is to predict the Closing Price of Yes Bank stock, Close is selected as the target variable. The Open, High, and Low prices are selected as predictor variables because they showed a strong positive relationship with the Closing Price during the Exploratory Data Analysis (EDA).

To improve the model's learning capability, additional features are created from the existing variables. These engineered features capture the stock's price movement and monthly volatility, providing more meaningful information to the regression models.

````python
df = data.copy()

# Create new features
df['Price_Range'] = df['High'] - df['Low']

df['Average_Price'] = (df['Open'] + df['High'] + df['Low']) / 3

df['Volatility'] = (df['High'] - df['Low']) / df['Open']

# Select independent variables
X = df[['Open',
        'High',
        'Low',
        'Price_Range',
        'Average_Price',
        'Volatility']]

# Target variable
y = df['Close']

# Display first few rows
X.head()
````

###Step 2: Data Preparation & Pre-processing

After selecting the features, the data is prepared for model training.

Since stock market data is time-series data, the dataset is first sorted by date. The data is then divided into 80% training data and 20% testing data. Finally, StandardScaler is applied to standardize the numerical features so that algorithms sensitive to feature scale perform efficiently.

````python
df = df.sort_values('Date')

# Re-create feature matrix after sorting
X = df[['Open',
        'High',
        'Low',
        'Price_Range',
        'Average_Price',
        'Volatility']]

y = df['Close']

# Chronological Train-Test Split
split = int(len(df) * 0.80)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

print("Training Shape :", X_train.shape)
print("Testing Shape  :", X_test.shape)
````

###Step 3: Data Validation Before Model Building

Before training the regression models, the processed data is validated to ensure that it is suitable for machine learning.

The correlation between variables is examined to identify strong relationships and detect multicollinearity. Since the EDA showed a high correlation between Open, High, Low, and Close, these variables are retained for model building.
