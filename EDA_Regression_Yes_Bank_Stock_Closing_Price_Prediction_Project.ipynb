## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Create a copy of the original dataset- so thatany changes made during data preprocessing do not affect the original data.
data = df.copy()
````

````python
# Convert Date column into datetime format makes it easier to sort the data and perform time-based analysis.
data['Date'] = pd.to_datetime(data['Date'], format='%b-%y')

# Check data types
data.dtypes
````

````python
# Sort dataset by Date
# arranging the records in chronological order helps in understanding trends and is important for time-series analysis.
data = data.sort_values('Date')

# Reset index
data.reset_index(drop=True, inplace=True)

# Display first 5 rows
data.head()
````

````python
# Check missing values - missing value can affect the accuracy of analysis so this is forcheck the missing value
data.isnull().sum()
````

````python
# Check duplicate rows
# Duplicate records may lead to biased analysis and incorrect model predictions.
# We check whether any duplicate rows exist.
data.duplicated().sum()
````

````python
# Remove duplicates dynamically if present
if data.duplicated().sum() > 0:
    data.drop_duplicates(inplace=True)

# Operational Rule Check: Validating financial pricing logic
is_high_valid = (data['High'] >= data['Open']).all() and (data['High'] >= data['Close']).all() and (data['High'] >= data['Low']).all()
is_low_valid = (data['Low'] <= data['Open']).all() and (data['Low'] <= data['Close']).all()

print(f"Data Consistency Check passed: High validation -> {is_high_valid} | Low validation -> {is_low_valid}")
````

````python
# sort the data set
data = data.sort_values('Date')

data.reset_index(drop=True, inplace=True)
````

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

### Chart 1: Interactive Historical Candlestick Chart (OHLC Analysis)

````python
fig = go.Figure(data=[go.Candlestick(
    x=data['Date'],
    open=data['Open'],
    high=data['High'],
    low=data['Low'],
    close=data['Close'],
    increasing_line_color='green',
    decreasing_line_color='red'
)])

fig.update_layout(
    title='Yes Bank Historical Monthly OHLC Trend',
    xaxis_title='Timeline',
    yaxis_title='Price (₹)',
    xaxis_rangeslider_visible=False,
    template='plotly_white',
    width=950,
    height=500
)
fig.show()
````

Why this chart? Candlestick charts capture price ranges within a defined period, detailing trading velocity and direction.

Insights: Highlights a clear structural paradigm break post-2018, shifting from a long-term bullish run to a deep capitulation window.

Business Impact: Identifies macro-regimes to prevent models from learning unrepresentative bull states without accounting for stress risks.

### Chart 2: Distribution and Spread (Box Plot of Closing Price)

````python
plt.figure(figsize=(7, 4))
sns.boxplot(x=data['Close'], color='#4C6EF5')
plt.title("Statistical Distribution of Closing Price", fontsize=12)
plt.xlabel("Price (₹)")
plt.tight_layout()
plt.show()
````

Why this chart? To expose historical skewness, median pricing, and potential pricing outliers.

### Chart 9: 6-Month Rolling Moving Average (Trend Smoothing)

````python
data['Rolling_Mean_6M'] = data['Close'].rolling(window=6).mean()

plt.figure(figsize=(10, 5))
plt.plot(data['Date'], data['Close'], label='Raw Closing Price', color='lightgray', alpha=0.7)
plt.plot(data['Date'], data['Rolling_Mean_6M'], color='#7048E8', linewidth=2.5, label='6-Month Moving Average')
plt.title("Closing Price Evolution & 6-Month Macro Smoothing", fontsize=12)
plt.xlabel("Timeline")
plt.ylabel("Price (₹)")
plt.legend()
plt.tight_layout()
plt.show()
````

Why this chart? Strips out transient monthly noise to focus purely on structural direction.

Insights: Explicitly tracks the steady loss of momentum heading directly into the corporate collapse of late 2018.
