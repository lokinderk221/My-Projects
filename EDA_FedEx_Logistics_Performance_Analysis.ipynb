## 3. ***Data Wrangling***

### Data Wrangling Code

````python
import pandas as pd
import numpy as np

dataset = pd.read_csv('/content/SCMS_Delivery_History_Dataset.csv')
df = dataset.copy()

df.drop_duplicates(inplace=True)

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
    .str.replace("#", "")
    .str.replace(r"[^\w]", "", regex=True)
)

str_cols = df.select_dtypes(include="object").columns
df[str_cols] = df[str_cols].apply(lambda x: x.str.strip())

date_cols = [
    "pq_first_sent_to_client_date",
    "po_sent_to_vendor_date",
    "scheduled_delivery_date",
    "delivered_to_client_date",
    "delivery_recorded_date"
]

for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

num_cols = [
    "line_item_quantity",
    "line_item_value",
    "pack_price",
    "unit_price",
    "weight_kilograms",
    "freight_cost_usd",
    "line_item_insurance_usd"
]

for col in num_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
        df[col].fillna(df[col].median(), inplace=True)

cat_cols = df.select_dtypes(include="object").columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

df["delivery_delay_days"] = (
    df["delivered_to_client_date"] - df["scheduled_delivery_date"]
).dt.days

df["delayed_flag"] = np.where(df["delivery_delay_days"] > 0, "Delayed", "On-Time")

df["delivery_month"] = df["delivered_to_client_date"].dt.to_period("M")

df.to_csv("cleaned_scms_delivery_dataset.csv", index=False)
````

````python
df["delayed_flag"] = (df["delivery_delay_days"] > 0).astype(int)
# Calculate delivery delay days and convert negative delays to 0
df["delivery_delay_days"] = (
    df["delivered_to_client_date"] - df["scheduled_delivery_date"]
).dt.days.clip(lower=0)

# Create delayed flag
df["delayed_flag"] = np.where(df["delivery_delay_days"] > 0, "Delayed", "On-Time")

````

Chart - 1 Distribution of Delivery Delay (Histogram / KDE Plot)

````python
# Chart - 1 visualization code

plt.figure(figsize=(10,6))

# Histogram with KDE
sns.histplot(
    df["delivery_delay_days"],  # column to plot
    bins=30,                    # number of bins
    kde=True,                   # overlay KDE curve
    color='skyblue'             # color of histogram
)

# Add labels and title
plt.title("Distribution of Delivery Delay Days", fontsize=16)
plt.xlabel("Delivery Delay (Days)", fontsize=14)
plt.ylabel("Number of Deliveries", fontsize=14)

# Add a vertical line at mean for reference
plt.axvline(df["delivery_delay_days"].mean(), color='red', linestyle='--', label=f"Mean: {df['delivery_delay_days'].mean():.1f}")
plt.legend()

# Optional: grid for better readability
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()

#This histogram shows the frequency distribution of delivery delays. Bar labels are added to clearly indicate the number of shipments falling within each delay range.
````
